# Export cleaned CSV

**Purpose:** write `master_customer_behavior_clean.csv` using the **same rules** as descriptive work, via **`shared_cleaning_rules.py`**.

**Why a separate notebook?** One place to rebuild the clean file so the main notebook does not overwrite it every time you *Run all*.

**After running:** refresh the Streamlit app if it reads the clean CSV.

```python
from pathlib import Path
import sys

# Repo root: cwd may be project root or descriptive/ (one level down).
_repo = Path.cwd().resolve()
if not (_repo / "shared_cleaning_rules.py").is_file():
    _repo = _repo.parent
if not (_repo / "shared_cleaning_rules.py").is_file():
    raise RuntimeError(
        "Could not find shared_cleaning_rules.py. "
        "Set the notebook working directory to the project root or descriptive/."
    )

if str(_repo) not in sys.path:
    sys.path.insert(0, str(_repo))

from shared_cleaning_rules import export_clean_csv

out = export_clean_csv(project_root=_repo)
print(f"Wrote: {out}")
print("Streamlit: streamlit run streamlit_dashboard_app.py")
```
